```python
import sys
sys.path.append('../../..')
```

```python
is_hitter = True
```

```python
from Model.Constants import GetDataPrepBinaryFile
from Model.Combined.DataPrep.Data_Prep import Combined_Data_Prep

data_prep = Combined_Data_Prep.Load_From_File("../../" + GetDataPrepBinaryFile(1))
```

```python
io_list = data_prep.Generate_IO_Hitters(is_training=True) if is_hitter else data_prep.Generate_IO_Pitchers(is_training=True)
```

```python
from Model.Combined.Tuning.ProTuningShared import SearchWidth
from Model.Combined.Tuning.ProHead import trial_runner, ProModelHeadTuningRecipe

recipes = [ProModelHeadTuningRecipe.PA,
           ProModelHeadTuningRecipe.POS,
           ProModelHeadTuningRecipe.PT,
           ProModelHeadTuningRecipe.LEVEL,
           ProModelHeadTuningRecipe.STATS,
           ProModelHeadTuningRecipe.MLBVALUE,
           ProModelHeadTuningRecipe.MLBSTAT]
width = SearchWidth.WIDE
```

```python
studies = trial_runner(
    io_list=io_list,
    data_prep=data_prep,
    is_hitter=is_hitter,
    num_trials=100,
    width=width,
    recipes=recipes,
)
```

```python
for i, study in enumerate(studies):
    print(recipes[i].name)
    print(f"Final best value after refinement: {study.best_value:.4f}")
    print(f"Best hyperparameters: {study.best_params}\n\n\n")
```

```python
import optuna.visualization as vis
for study in studies:
    vis.plot_param_importances(study).show()
```
